```python
#Q1. Phone Contact Book
contacts = {}

while True:
    print("1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter name: ")
        if name in contacts:
            print("Contact already exists")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added")
    elif choice == "2":
        name = input("Enter name to search: ")
        if name in contacts:
            print(name, ":", contacts[name])
        else:
            print("Contact not found")
    elif choice == "3":
        for name, phone in contacts.items():
            print(name, ":", phone)
    elif choice == "4":
        break
    else:
        print("Invalid choice")
```

```python
#Q3. Student Marks Dictionary
marks = {"Aman": 82, "Priya": 91, "Rahul": 45, "Sneha": 67, "Kiran": 30}

for name, m in marks.items():
    if m >= 40:
        print(name, "- Pass")
    else:
        print(name, "- Fail")

topper = ""
top = 0
for name, m in marks.items():
    if m > top:
        top = m
        topper = name
print("Topper:", topper, top)

passed = {}
for name, m in marks.items():
    if m >= 40:
        passed[name] = m
print("Passed students:", passed)
```

```python
#Q4. Common Friends
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

print("Mutual friends:", riya_friends.intersection(arjun_friends))
print("Only Riya's friends:", riya_friends.difference(arjun_friends))
print("Everyone:", riya_friends.union(arjun_friends))
print("Suggestions for Riya:", arjun_friends.difference(riya_friends))
```

```python
#Q5. Character Counter
text = input("Enter text: ")
counts = {}

for ch in text:
    if ch != " ":
        counts[ch] = counts.get(ch, 0) + 1

print(counts)

most = ""
highest = 0
for ch, c in counts.items():
    if c > highest:
        highest = c
        most = ch
print("Most frequent character:", most, "(" + str(highest) + " times)")
```

```python
#Q6. Restaurant Billing
menu = {"dosa": 60, "idli": 40, "vada": 30, "coffee": 25, "tea": 15}
total = 0

while True:
    item = input("What would you like to order? ").lower()
    if item == "bill":
        break
    if item in menu:
        total = total + menu[item]
        print(item, "added")
    else:
        print("Item not available")

print("Total:", total)
if total > 200:
    total = total - total * 10 / 100
    print("10% discount applied. Final bill:", total)
```

```python
#Q7. City Coordinates
cities = [("Hyderabad", 17.38, 78.48),
          ("Mumbai", 19.07, 72.87),
          ("Chennai", 13.08, 80.27),
          ("Delhi", 28.61, 77.20)]

for name, lat, lon in cities:
    print(name, "- Latitude:", lat, "Longitude:", lon)

north_city = ""
max_lat = 0
for name, lat, lon in cities:
    if lat > max_lat:
        max_lat = lat
        north_city = name
print("Farthest north:", north_city, max_lat)
```

```python
#Q8. Library Book Tracker
books = {"Python Basics": 3, "Data Science 101": 0,
         "Machine Learning": 2, "SQL Guide": 1}

book = input("Enter book name: ")

if book not in books:
    print("Book not found")
elif books[book] == 0:
    print("Currently unavailable")
else:
    books[book] = books[book] - 1
    print("Book issued successfully")

print(books)
```

```python
#Q9. Group Employees by Department
employees = [("Ravi", "IT"), ("Anjali", "HR"), ("Suresh", "IT"),
             ("Meena", "Finance"), ("Arun", "HR"), ("Divya", "IT")]

dept = {}
for name, d in employees:
    if d not in dept:
        dept[d] = []
    dept[d].append(name)

print(dept)
for d, names in dept.items():
    print(d, ":", len(names), "employees")
```

```python
#Q10. Quiz Answer Checker
answer_key = ("B", "C", "A", "D", "B")
score = 0
wrong = set()

for i in range(5):
    ans = input("Answer for Q" + str(i + 1) + ": ").upper()
    if ans == answer_key[i]:
        print("Q" + str(i + 1), "Correct")
        score = score + 1
    else:
        print("Q" + str(i + 1), "Wrong")
        wrong.add(i + 1)

print("Wrong questions:", wrong)
print("Score:", score, "/ 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
