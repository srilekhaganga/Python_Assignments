```python
#Q1. Class Marks Report
marks = [78, 45, 92, 33, 67, 88, 51, 29]

print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Average:", sum(marks) / len(marks))

passed = 0
for m in marks:
    if m >= 35:
        passed = passed + 1
print("Passed:", passed)
```

```python
#Q2. Grocery Shopping List
items = []

while True:
    item = input("Enter item: ")
    if item == "done":
        break
    if item in items:
        print("Already added")
    else:
        items.append(item)

for i in range(len(items)):
    print(str(i + 1) + ". " + items[i])
print("Total items:", len(items))
```

```python
#Q3. Username Validator
name = input("Enter username: ")

if len(name) < 5 or len(name) > 12:
    print("Invalid: Username must be 5 to 12 characters")
elif " " in name:
    print("Invalid: Username cannot have spaces")
elif name[0].isdigit():
    print("Invalid: Username cannot start with a digit")
else:
    print("Valid username")
```

```python
#Q4. Vowel Counter for a Tweet
text = input("Enter a sentence: ").lower()
vowels = 0
consonants = 0
spaces = 0

for ch in text:
    if ch in "aeiou":
        vowels = vowels + 1
    elif ch.isalpha():
        consonants = consonants + 1
    elif ch == " ":
        spaces = spaces + 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

```python
#Q5. Palindrome Name Checker
name = input("Enter your name: ").lower()

if name == name[::-1]:
    print("Bonus! Your name is a palindrome.")
else:
    print("Not a palindrome.")
```

```python
#Q6. Temperature Tracker
temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print("Heat alert days:")
for i in range(len(temps)):
    if temps[i] > 37:
        print(days[i])

hottest = max(temps)
print("Hottest day:", days[temps.index(hottest)], hottest)

fahrenheit = []
for t in temps:
    fahrenheit.append(t * 9 / 5 + 32)
print("Fahrenheit:", fahrenheit)
```

```python
#Q7. Word Frequency in a Sentence
sentence = input("Enter a sentence: ")
words = sentence.split()

longest = words[0]
shortest = words[0]
capital = 0

for w in words:
    if len(w) > len(longest):
        longest = w
    if len(w) < len(shortest):
        shortest = w
    if w[0].isupper():
        capital = capital + 1

print("Longest word:", longest)
print("Shortest word:", shortest)
print("Words starting with capital letter:", capital)
```

```python
#Q8. Positive and Negative Balances
balances = [1500, -200, 3000, -50, 0, 750, -1000]
positive_accounts = []
negative_accounts = []

for b in balances:
    if b > 0:
        positive_accounts.append(b)
    elif b < 0:
        negative_accounts.append(b)

print("Positive:", positive_accounts)
print("Negative:", negative_accounts)
print("Total owed:", -sum(negative_accounts))
```

```python
#Q9. Password Strength
password = input("Enter password: ")
has_upper = False
has_lower = False
has_digit = False

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True

count = 0
if has_upper:
    count = count + 1
if has_lower:
    count = count + 1
if has_digit:
    count = count + 1
if len(password) >= 8:
    count = count + 1

if count == 4:
    print("Strong")
elif count >= 2:
    print("Medium")
else:
    print("Weak")
```
