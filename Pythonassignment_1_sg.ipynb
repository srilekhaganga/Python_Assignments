```python
#Multiplication Table for a Kid
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8

print("Total bill: Rs.", bill)
```

```python
#Savings Piggy Bank
total = 0
days = 0

while total < 1000:
    saving = int(input("Enter today's savings: "))
    total = total + saving
    days = days + 1

print("Goal reached in", days, "days! Total saved: Rs.", total)
```

```python
#Q5. ATM PIN Check
correct_pin = 4321
attempts = 0

while attempts < 3:
    pin = int(input("Enter PIN: "))
    if pin == correct_pin:
        print("Access granted")
        break
    else:
        print("Wrong PIN")
        attempts = attempts + 1

if attempts == 3:
    print("Card blocked")
```

```python
6 #Even and Odd Counter
n = int(input("Enter N: "))
even = 0
odd = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
# Q7. Number Guessing Game
secret = 27
guess = 0
count = 0

while guess != secret:
    guess = int(input("Guess the number: "))
    count = count + 1
    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")

print("Correct! You took", count, "guesses.")
```

```python
#Q8. Exam Result with Grade
marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Fail")
```

```python
#Q9. Sum of Digits on a Bill Number
n = int(input("Enter bill number: "))
total = 0

while n > 0:
    total = total + n % 10
    n = n // 10

print("Sum of digits:", total)

if total > 20:
    print("Congratulations! You win a free gift.")
else:
    print("Sorry, no gift this time.")
```

```python
#Q10. Staircase Pattern
steps = int(input("Enter number of steps: "))

for i in range(1, steps + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```
